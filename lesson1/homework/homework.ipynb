```python
import numpy as np
def weighted_mean(values,weights):
    # 使用numpy的dot函数计算加权平均值
    try:
        if sum(weights)==1 and len(values)==len(weights):
            all_sum=np.dot(values,weights)
            all_weight=np.sum(weights)
            return all_sum/all_weight
        else:
            raise ValueError("weights must be a list of numbers and sum to 1")
    except ValueError as e:
        print(e)

weighted_mean([1, 2, 3], [0.2, 0.3, 0.5]) 
```

```python
def topk(items,k):
    ranked=sorted(items,key=lambda x:x[1],reverse=True)
    catch_k=ranked[:k]
    answer_complete=sorted(catch_k,key=lambda x:x[0])
    answer_real=[name for name,score in answer_complete]
    return answer_real
topk([("b", 90), ("a", 90), ("c", 80)], 2)
```

```python
from typing import Sequence
from typing import List
def standardize(xs:Sequence[float],ddot:int=1) -> List[float]:
    mean=np.mean(xs)
    std_0=np.std(xs)
    std_1=np.std(xs,ddof=1)
    return [(xs-mean)/std_0,(xs-mean)/std_1]
standardize([1,2,3,4,])
```
