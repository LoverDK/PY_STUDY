```python
import pandas as pd

codes = pd.Series(["PY","SQL","X",None], index=["s1","s2","s3","s4"], dtype="object")
names = codes.map({"PY":"Python","SQL":"SQL"})
print(names)
# s1 Python；s2 SQL；s3/s4 为缺失
```

```python
mapping = {"PY":"Python","SQL":"SQL"}
def course_name(value):
    if pd.isna(value):
        return "Missing"
    return mapping.get(value, "Unknown")

print(codes.map(course_name).tolist())
# ["Python","SQL","Unknown","Missing"]
```

```python
def grade(value):
    if pd.isna(value):
        return "Missing"
    if value >= 90:
        return "Excellent"
    if value >= 60:
        return "Pass"
    return "Fail"

scores = pd.Series([59,60,89,90,None], dtype="float64")
print(scores.map(grade).tolist())
print(scores.apply(grade).tolist())
# 两者在本例均为 ["Fail","Pass","Pass","Excellent","Missing"]
```

```python
orders = pd.DataFrame(
    {"price":[10.0,20.0,None], "qty":[2,3,1]},
    index=["o3","o1","o8"],
)
print(orders)
result = orders.assign(amount=lambda d: d["price"]*d["qty"])
print(result["amount"].tolist()) # 20.0、60.0、缺失
```

```python
mapping={'PY':'Python','SQL':'SQL','C':'C'}
def mis(value):
    if pd.isna(value):
        return 'Missing'
    elif value not in mapping:
        return 'Unknown'
    else:
        return mapping[value]
def label_courses(codes: pd.Series) -> pd.Series:
    return codes.map(mis)
codes = pd.Series(["PY","SQL","X",None,"py",""], index=[9,3,7,2,8,4], dtype="object")
# 输出 ["Python","SQL","Unknown","Missing","Unknown","Unknown"]
label_courses(codes)
```

```python
def grade(x):
    if pd.isna(x):
        return 'Missing'
    elif x>=90:
        return 'Excellent'
    elif x>=60:
        return 'Pass'
    else:
        return 'Fail'
def add_score_band(df: pd.DataFrame) -> pd.DataFrame:
    s=df['score']
    bad=s.notna()&((s<0)|(s>100))
    if bad.any():
        raise ValueError("score 必须在 0 到 100 之间")
    return df.assign(bond=s.map(grade))
```

```python
import numpy as np
def add_order_totals(df: pd.DataFrame) -> pd.DataFrame:
    s=df[['price','qty','discount']]
    bad=s.notna()&(s<0)
    if bad.any().any():
        raise ValueError("订单价格不能为负")
    df_2=df.assign(net=np.maximum(df['price']*df['qty']-df['discount'],0.0))
    df_3=df.assign(shipping=df_2.apply(lambda s:0.0 if df_2['net']>=100 else 8.0,axis=1))
    return df_3
```
