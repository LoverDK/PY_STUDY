```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x=np.array([1,2,3,4])
y=np.array([2,3,4,5])
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('plot')
plt.grid(True)
plt.savefig('plot.png')
```

```python
x=np.arange(1,8)
y1 = np.array([10, 12, 11, 15, 14, 18, 17])
y2 = np.array([9, 10, 12, 11, 13, 14, 16])
plt.plot(x, y1, "b-o", label="store_A")       # 蓝实线+圆点
plt.plot(x, y2, "r--s", label="store_B")      # 红虚线+方点
plt.xlabel("Day")
plt.ylabel("Sales")
plt.title("Two stores")
plt.grid(True)
plt.legend()                                   # 多系列时看一眼即可；深讲留给 L07
```

```python
rng=np.random.default_rng(0)
x=rng.normal(0,1,50)
y=2*x+rng.normal(0,0.5,50)
sizes=np.abs(y)*30
plt.scatter(x,y,s=sizes,c=y,cmap='viridis',alpha=0.8)
plt.xlabel('x')
plt.ylabel('y')
plt.title('scatter')
plt.colorbar()
plt.grid(True)
```

```python
t = np.linspace(0, 2 * np.pi, 100)
plt.plot(t, np.sin(t), label="sin")
plt.plot(t, np.cos(t), label="cos")
plt.legend()
```

```python
def plot_series(x,y,out_path:str)->str:
    if len(x)!=len(y):
        raise ValueError("x and y must have the same length")
    fig=plt.figure()
    try:
        plt.plot(x,y,'b-o')
        xlabel='x'
        ylabel='y'
        title='series'
        plt.grid(True)
        plt.savefig(out_path,format='png')
    finally:
        plt.close(fig)
    return out_path
plot_series([1,2,3],[10,20,15],'hw6_1.png')     


```

```python
def plot_scatter(x,y,out_path:str)->str:
    if len(x)!=len(y):
        raise ValueError("x and y must have the same length")
    fig=plt.figure()
    try:
        plt.scatter(x,y)
        xlabel='x'
        ylabel='y'
        title='scatter'
        plt.grid(True)
        plt.savefig(out_path,format='png')
    finally:
        plt.close(fig)
    return out_path
plot_scatter([1.0,2.0,3.0,4.0],[2.1,3.9,6.2,7.8],'hw6_2.png')
```

```python
def all_equal(a,b,c):
    if not(a==b and a==c):
        return True
    else:
        return False
def plot_two_series(x,y1,y2,out_path:str)->str:
    if all_equal(len(x),len(y1),len(y2)):
        raise ValueError("x,y1,y2 must have the same length")
    fig=plt.figure()
    try:
        plt.plot(x,y1,'b-o',label='A')
        plt.plot(x,y2,'r--s',label='B')
        xlabel='x'
        ylabel='y'
        title='A vs B'
        plt.grid(True)
        plt.legend()
        plt.savefig(out_path,format='png')
    finally:
        plt.close(fig)
    return out_path
plot_two_series(
    [1,2,3,4],
    [10,11,13,12],
    [9,10,10,12],
    'hw6_3.png'
)
```
