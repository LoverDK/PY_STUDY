```python
def reset_by_two(n)->int:
    num=0
    for i in range(n):
        j=0
        while j<=i:
            j+=2
            num+=1
    return num
assert reset_by_two(0) == 0
assert reset_by_two(1) == 1
assert reset_by_two(2) == 2
assert reset_by_two(3) == 4
assert reset_by_two(5) == 9
assert reset_by_two(6) == 12
for n in range(41):
    assert reset_by_two(n) == ((n + 1) ** 2) // 4  
```

```python
def shared_pointer(n)->tuple[int,int]:
    n_move=0
    n_check=0
    j=0
    for i in range(n):
        while True:
            if (j>=i+2):
                n_check+=1
                break
            else:
                j+=1
                n_check+=1
                n_move+=1
    return (n_move,n_check)
assert shared_pointer(0) == (0, 0)
assert shared_pointer(1) == (2, 3)
assert shared_pointer(2) == (3, 5)
assert shared_pointer(5) == (6, 11)
for n in range(1, 51):
    assert shared_pointer(n) == (n + 1, 2 * n + 1)
```

```python
def append_costs(m)->tuple[int,int,int,int]:
    n_append=0
    n_copy=0
    total_append=0
    total_copy=0
    total_cost=0
    max_cost=0
    current_cost=0
    current_capacity=1
    current_length=0
    while current_length<m:
        if current_length<current_capacity:
            current_length+=1
            n_append+=1
        else:
            current_capacity*=2
            n_copy+=current_length
            current_length+=1
            n_append+=1
        current_cost=n_append+n_copy
        total_append+=n_append
        total_copy+=n_copy
        n_append=0
        n_copy=0
        total_cost+=current_cost
        if current_cost>max_cost:
            max_cost=current_cost
        if current_length==m:
            return (total_append,total_copy,total_cost,max_cost)
    return (total_append,total_copy,total_cost,max_cost)
assert append_costs(0) == (0, 0, 0, 0)
assert append_costs(1) == (1, 0, 1, 1)
assert append_costs(2) == (2, 1, 3, 2)
assert append_costs(4) == (4, 3, 7, 3)
assert append_costs(5) == (5, 7, 12, 5)
assert append_costs(8) == (8, 7, 15, 5)
assert append_costs(9) == (9, 15, 24, 9)
for m in range(1, 101):
    writes, copies, total, peak = append_costs(m)
    assert writes == m
    assert copies == 2 ** ((m - 1).bit_length()) - 1
    assert total == writes + copies
    assert m <= total < 3 * m
    assert 1 <= peak <= total


```
