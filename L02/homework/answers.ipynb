## 练习 1 阶乘


```python
def fact(n):
    r = 1
    for i in range(2, n + 1):
        r *= i
    return r

# n = int(input())
# print(fact(n))
print(fact(5), fact(0))
```

## 练习 2 非空行


```python
def count_nonempty(lines):
    return sum(1 for ln in lines if ln.strip())

# m = int(input())
# lines = [input() for _ in range(m)]
# print(count_nonempty(lines))
demo = ["hello", "", "  world", "   "]
print(count_nonempty(demo))
```

## 练习 3 Student


```python
class Person:
    def __init__(self, name):
        self.name = name
    def hello(self):
        return f"I am {self.name}"

class Student(Person):
    def __init__(self, name, sid):
        super().__init__(name)
        self.sid = sid
    def hello(self):
        return f"Student {self.name} id={self.sid}"

# name, sid = input().split()
# print(Student(name, sid).hello())
print(Student("Qiu", "2026001").hello())
```
