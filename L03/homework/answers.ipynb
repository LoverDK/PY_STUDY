## 练习 1 销售表摸底

```python
import pandas as pd

def sales_overview(csv_path: str) -> tuple[int, int, int]:
    df = pd.read_csv(csv_path)
    nrows, ncols = df.shape
    revenue_max = int(df["revenue"].max())
    return (nrows, ncols, revenue_max)

print(sales_overview("sales.csv"))
# 期望: (10, 5, 35988)
```

## 练习 2 学生及格名单

```python
def passed_names(json_path: str) -> list[str]:
    df = pd.read_json(json_path)
    return (
        df[df["score"] >= 60]
        .sort_values("score", ascending=False)["name"]
        .tolist()
    )

print(passed_names("students.json"))
# 期望: ['周八', '张三', '陈一', '孙七', '李四', '赵五']
```

## 练习 3 区域高销产品

```python
def high_revenue_products(csv_path: str, region: str, threshold: int) -> list[str]:
    df = pd.read_csv(csv_path)
    mask = (df["region"] == region) & (df["revenue"] >= threshold)
    return df.loc[mask, "product"].tolist()  # 保持原表行序，不去重不排序

print(high_revenue_products("sales.csv", "华东", 5000))
# 期望: ['笔记本', '键盘']
print(high_revenue_products("sales.csv", "华北", 10000))
# 期望: ['笔记本']
```
