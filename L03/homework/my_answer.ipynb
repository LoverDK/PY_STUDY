```python
import pandas as pd
def sales_overview(csv_path:str)->tuple[int,int,int]:
    df=pd.read_csv(csv_path)
    return (df.shape[0],df.shape[1],df['revenue'].max())
sales_overview('sales.csv')
```

```python
def passwd_names(json_path:str)->list[str]:
    df=pd.read_json(json_path)
    df_good=df[df['score']>=60].sort_values('score',ascending=False)
    return df_good['name'].tolist()
passwd_names('students.json')
```

```python
def high_revenue_products(csv_path:str,region:str,threshold:int)->list[str]:
    df=pd.read_csv(csv_path)
    df_good=df[(df['region']==region) & (df['revenue']>=threshold)]
    return df_good['product'].tolist()
high_revenue_products('sales.csv','华东',5000)
```
