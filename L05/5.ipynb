```python
def filled_score_mean(csv_path:str)->float:
    df=pd.read_csv(csv_path)
    df_1=df.dropna(subset='score')
    df=df.fillna(df_1['score'].mean())
    return df['score'].mean()
filled_score_mean('scores.csv')
```

```python
def major_stats(df:'pd.DataFrame')->'pd.Dataframe':
    df_1=df.dropna(subset=['score'])
    return df_1.groupby('major').agg(
        avg_score=('score','mean'),
        n=('score','count'),
        total_credits=('score','sum')
    )
df=pd.read_csv('scores.csv')
major_stats(df)
```

```python
def paid_amount_by_city(orders_csv:str,customers_csv:str)->dict[str,float]:
    orders=pd.read_csv(orders_csv)
    customers=pd.read_csv(customers_csv)
    orders=orders.fillna(0)
    all=pd.merge(orders,customers,on='customer_id',how='left')
    all['city']=all['city'].fillna('UNKNOWN')
    all=all[all['status']=='paid']
    answer=all.groupby('city')['amount'].sum()
    return answer.to_dict()
paid_amount_by_city('orders.csv','customers.csv')
```
