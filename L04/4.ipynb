```python
import numpy as np
def row_sum(n_rows:int , n_cols:int )->list[int]:
    A=np.arange(0,n_rows*n_cols).reshape(n_rows,n_cols)
    return np.sum(A,axis=1).tolist()
row_sum(3,4)
```

```python
def passed_mean(scores:list[float],threshold:float=60.0)->str:
    score=np.array(scores)
    mask=score>=threshold
    if score[mask].shape[0]>0:
        passed_mean=score[mask].mean()
        return str(passed_mean)
    else:
        return 'NA'
passed_mean([55, 72, 91, 60, 48])
passed_mean([40, 50, 59], 60)
passed_mean([60, 60], 60)
```

```python
def center_columns(data:list[list[float]])->list[list[float]]:
    dt=np.array(data)
    dt_mean=dt.mean(axis=0)
    ans=dt-dt_mean
    return ans.tolist()
center_columns([[1, 2, 3], [4, 5, 6]])
```
