## 练习 1 reshape 与按行求和

```python
import numpy as np

def row_sums(n_rows: int, n_cols: int) -> list[int]:
    arr = np.arange(n_rows * n_cols).reshape(n_rows, n_cols)
    return arr.sum(axis=1).astype(int).tolist()

print(row_sums(3, 4))  # 期望: [6, 22, 38]
print(row_sums(2, 3))  # 期望: [3, 12]
```

## 练习 2 布尔筛选后求均值

```python
def passed_mean(scores: list[float], threshold: float = 60.0) -> str:
    arr = np.asarray(scores, dtype=float)
    passed = arr[arr >= threshold]
    if passed.size == 0:
        return "NA"
    return f"{passed.mean():.2f}"

print(passed_mean([55, 72, 91, 60, 48]))  # 期望: "74.33"
print(passed_mean([40, 50, 59], 60))      # 期望: "NA"
print(passed_mean([60, 60], 60))          # 期望: "60.00"
```

## 练习 3 广播：每列减均值

```python
def center_columns(data: list[list[float]]) -> list[list[float]]:
    X = np.asarray(data, dtype=float)
    return (X - X.mean(axis=0)).tolist()

print(center_columns([[1, 2, 3], [4, 5, 6]]))
# 期望: [[-1.5, -1.5, -1.5], [1.5, 1.5, 1.5]]
```
