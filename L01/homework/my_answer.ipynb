```python
n=int(input())
nums=list(map(int,input().split()))
grades=[]
for num in nums:
    if num>=90:
        grades.append("A")
    elif num>=60:
        grades.append("B")
    else:
        grades.append("C")
print(" ".join(grades))
```

```python
sentence=input().split()
ss=set(sentence)
print(len(ss))
cnt={word:sentence.count(word) for word in ss}
for word in sorted(cnt):
    print(word,cnt[word])
```
