## 练习 1：成绩分档

**题意**：第一行整数 `n`，第二行 `n` 个分数。按规则输出等级（空格分隔）：
- 分数 ≥ 90 → `A`
- 分数 ≥ 60 → `B`
- 否则 → `C`

**样例输入**
```
4
95 60 59 90
```

**样例输出**
```
A B C A
```


````python
# 机试写法：用 input()
# n = int(input())
# scores = list(map(int, input().split()))

# Jupyter 演示：用样例字符串模拟标准输入
from io import StringIO
import sys

sys.stdin = StringIO("4\n95 60 59 90\n")

n = int(input())
scores = list(map(int, input().split()))
assert len(scores) == n

grades = []
for s in scores:
    if s >= 90:
        grades.append("A")
    elif s >= 60:
        grades.append("B")
    else:
        grades.append("C")

print(" ".join(grades))
# 期望：A B C A
````

````python
# 等价写法：列表推导式 + 条件表达式（与「改值推导」同形）
scores = [95, 60, 59, 90]
grades = ["A" if s >= 90 else "B" if s >= 60 else "C" for s in scores]
print(" ".join(grades))
````

## 练习 2：词频

**题意**：读入一行单词（空格分隔）。
1. 先输出不同单词的个数；
2. 再按**字典序**逐行输出 `单词 次数`。

**样例输入**
```
ml ai ml stat ai pandas
```

**样例输出（示意）**
```
4
ai 2
ml 2
pandas 1
stat 1
```


````python
from io import StringIO
import sys

sys.stdin = StringIO("ml ai ml stat ai pandas\n")

words = input().split()
cnt = {}
for w in words:
    cnt[w] = cnt.get(w, 0) + 1

print(len(cnt))
for w in sorted(cnt):
    print(w, cnt[w])
````

### 易错点
- 先 `print(len(cnt))`，再按 `sorted(cnt)` 输出，不要边计边打乱顺序。
- 字典的插入顺序 ≠ 字典序；必须 `sorted`。
- 不要用 `set` 只算个数却丢频率：本题要次数。


## 练习 3：正偶数的平方（必须用列表推导式）

**题意**：读入一行整数，输出其中所有**正偶数**的平方，空格分隔。若没有符合条件的数，输出空行即可。

**要求**：代码中必须出现列表推导式。

**样例输入**
```
-2 3 4 0 6 7
```

**样例输出**
```
16 36
```

说明：`4` 与 `6` 为正偶数；`-2` 不是正数；`0` 不是正数。


````python
from io import StringIO
import sys

sys.stdin = StringIO("-2 3 4 0 6 7\n")

nums = list(map(int, input().split()))
# 过滤推导：for 后面跟 if（没有 else）
squares = [x * x for x in nums if x > 0 and x % 2 == 0]
print(" ".join(map(str, squares)))
# 期望：16 36
````

### 易错点
- 「正偶数」= `x > 0` **且** `x % 2 == 0`；不要漏掉对 `0`、负数的排除。
- 过滤用 `[expr for x in xs if cond]`；不要写成带 else 的改值形式。
- 输出要用空格连接；空列表时 `" ".join([])` 得到空字符串，打印后是空行，符合题意。

````python
# 例2：出现次数 ≥2 的单词，字典序，空格分隔；没有则 NONE
from io import StringIO
import sys
sys.stdin = StringIO("ml ai ml stat ai\n")
words = input().split()
cnt = {}
for w in words:
    cnt[w] = cnt.get(w, 0) + 1
ans = sorted([w for w in cnt if cnt[w] >= 2])
print(" ".join(ans) if ans else "NONE")
````
